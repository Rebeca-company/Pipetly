# References Section Assessment

This notebook applies `_extract_references_section` to every paper in Step 6 and prints the extracted references section for manual validation.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / "models").exists():
    repo_root = repo_root.parent

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from models.paper import Paper
from processors.protocol_extractor import _extract_references_section
from utils.intermediate_io import STEP5_FILE, load_model_list

papers = load_model_list(STEP5_FILE, Paper)
print(f"Loaded {len(papers)} papers from intermediate_outputs/{STEP5_FILE}")
```

```python
print_empty = True

for idx, paper in enumerate(papers, start=1):
    full_text = (paper.full_text.content if paper.full_text else "") or ""
    references_section = _extract_references_section(full_text)

    print("=" * 120)
    print(f"[{idx}/{len(papers)}] {paper.title or 'Untitled'}")
    print(f"DOI: {paper.doi or 'NO DOI'}")
    print("-" * 120)

    if references_section:
        print(references_section)
    elif print_empty:
        print("(No references section found)")

    print()
```

```python
found = 0

for paper in papers:
    full_text = (paper.full_text.content if paper.full_text else "") or ""
    if _extract_references_section(full_text):
        found += 1

ratio = (found / len(papers)) if papers else 0
print(f"References section found in {found}/{len(papers)} papers ({ratio:.1%}).")
```
